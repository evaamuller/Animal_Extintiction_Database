```python
%%sql
CREATE TABLE Diseases (
    disease_id INT PRIMARY KEY AUTO_INCREMENT, ## if adding id as PK (better)
    disease_name VARCHAR(100) NOT NULL UNIQUE,
    disease_type VARCHAR(100)
);
```

```python
%%sql
CREATE TABLE Disease_Outbreaks(
    outbreak_id INT PRIMARY KEY AUTO_INCREMENT,
    disease_id INT NOT NULL, ## if changed PK of Diseases
    animal_id INT NOT NULL, 
    location_id INT NOT NULL, ## if changed PK of Location
    outbreak_year YEAR,
    disease_mortality DECIMAL(4,2), ## mortality is percentage so better decimal(like 45,50%) than int? 

    ## naming the constraints?
    FOREIGN KEY (disease_id) REFERENCES Diseases(disease_id),
    FOREIGN KEY (animal_id) REFERNECES Animal(animal_id),
    FOREIGN KEY (location_id) REFERENCES Location(location_id)
);
```

```python
%%sql
CREATE TABLE Location_Of_Animal(
    animal_id INT NOT NULL,    
    location_id INT NOT NULL, ## if changing location_name to location_id 
    
    PRIMARY KEY (animal_id, location_id),
    FOREIGN KEY (animal_id) REFERENCES Animal(animal_id),
    FOREIGN KEY (location_id) REFERENCES Location(location_id)
);
```
